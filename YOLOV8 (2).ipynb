```python
import os

img_path = '/content/train/images'
lbl_path = '/content/train/labels'

# Remove images without a corresponding label file or with an empty label file.
for file_name in os.listdir(img_path):
    if file_name.endswith('.jpg'):
        base = os.path.splitext(file_name)[0]
        label_file = base + '.txt'
        label_full_path = os.path.join(lbl_path, label_file)
        image_full_path = os.path.join(img_path, file_name)

        # If label file is missing, remove the image.
        if not os.path.exists(label_full_path):
            os.remove(image_full_path)
            print(f"Removed image {file_name} because label file is missing.")
        # If label file exists but is empty, remove both image and label.
        elif os.path.getsize(label_full_path) == 0:
            os.remove(image_full_path)
            os.remove(label_full_path)
            print(f"Removed image {file_name} and its label {label_file} because the label file is empty.")

# Remove label files that have no corresponding image.
for file_name in os.listdir(lbl_path):
    if file_name.endswith('.txt'):
        base = os.path.splitext(file_name)[0]
        img_file = base + '.jpg'
        if not os.path.exists(os.path.join(img_path, img_file)):
            os.remove(os.path.join(lbl_path, file_name))
            print(f"Removed label {file_name} because its corresponding image file is missing.")

```
